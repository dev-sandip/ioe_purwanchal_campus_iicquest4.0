```python
import pandas as pd
```

```python
infile  = "/home/raghav/Work/ioe_purwanchal_campus_iicquest4.0/ml/data_cleaned/right_wrong.csv"
outfile = "/home/raghav/Work/ioe_purwanchal_campus_iicquest4.0/labeled.csv"

first = True
for chunk in pd.read_csv(infile, chunksize=500, encoding="utf-8"):
    # right words → label 0
    r = pd.DataFrame({"word": chunk["Right"].astype(str), "label": 0})
    # wrong words → label 1
    w = pd.DataFrame({"word": chunk["Wrong"].astype(str), "label": 1})

    out = pd.concat([r, w], ignore_index=True)

    out.to_csv(outfile,
               mode="w" if first else "a",
               header=first,
               index=False,
               encoding="utf-8")
    first = False
    del chunk, r, w, out

print("Done. Saved to", outfile)
```
