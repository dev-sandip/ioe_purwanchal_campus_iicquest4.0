```python
import pandas as pd

infile  = "/home/raghav/Work/ioe_purwanchal_campus_iicquest4.0/ml/notebooks/cleaned_data.csv"
outfile = "/home/raghav/Work/ioe_purwanchal_campus_iicquest4.0/ml/notebooks/tokens.csv"

first = True
batch_num = 0

for chunk in pd.read_csv(infile, usecols=['text'], chunksize=500):
    # tokenize just these 500 rows
    tokens = chunk['text'].dropna().str.split().explode()

    # save this batch (append after the first)
    tokens.to_csv(outfile, mode='w' if first else 'a', index=False, header=False)
    first = False

    # free this batch from RAM before the next 500
    del chunk, tokens

    batch_num += 1
    print(f"Done batch {batch_num} ({batch_num * 500} rows)")
```
