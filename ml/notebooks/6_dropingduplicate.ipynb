```python
import pandas as pd

path = "/home/raghav/Work/ioe_purwanchal_campus_iicquest4.0/labeled.csv"
df = pd.read_csv(path, encoding="utf-8")
df["word"] = df["word"].astype(str)

# remove ONLY single-letter words that are label 1 — keep everything else, duplicates and all
df = df[~((df["word"].str.len() == 1) & (df["label"] == 1))].reset_index(drop=True)

df.to_csv("/home/raghav/Work/ioe_purwanchal_campus_iicquest4.0/cleaned_labeled.csv",
          index=False, encoding="utf-8")

print("Single-letter label-1 left:", ((df["word"].str.len()==1) & (df["label"]==1)).sum())
print("Total rows:", len(df))
```
