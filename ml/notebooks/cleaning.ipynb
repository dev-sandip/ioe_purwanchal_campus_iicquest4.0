```python
import re
import pandas as pd
df=pd.read_csv("/home/raghav/Work/ioe_purwanchal_campus_iicquest4.0/combined.csv")
def clean_nepali(text):
    text = str(text)
    # keep Devanagari characters and spaces, drop everything else
    text = re.sub(r'[^\u0900-\u097F\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()   # tidy up extra spaces
    return text

df['text'] = df['text'].apply(clean_nepali)
```
