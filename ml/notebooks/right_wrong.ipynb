```python
import pandas as pd
import random
```

```python
infile  = "/home/raghav/Work/ioe_purwanchal_campus_iicquest4.0/ml/notebooks/tokens.csv"
outfile = "/home/raghav/Work/ioe_purwanchal_campus_iicquest4.0/right_wrong.csv"
```

```python
def shuffle_word(word):
    word = str(word)
    chars = list(word)
    if len(set(chars)) < 2:
        return word
    shuffled = word
    while shuffled == word:
        random.shuffle(chars)
        shuffled = "".join(chars)
    return shuffled
```

```python
first = True
batch_num = 0
```

```python


for chunk in pd.read_csv(infile, header=None, names=["Right"], chunksize=500):
    # build the wrong column for these 500 words
    chunk["Wrong"] = chunk["Right"].apply(shuffle_word)

    # save this batch (write header only on the first)
    chunk.to_csv(outfile,
                 mode="w" if first else "a",
                 header=first,
                 index=False,
                 encoding="utf-8")
    first = False

    del chunk                      # free this batch from RAM
    batch_num += 1
    print(f"Done batch {batch_num} ({batch_num * 500} words)")
```
